First, we just test a few sample bm25 k values, the default b value for document length normalization is 0.75

```python
importlib.reload(recsys_ii)
importlib.reload(inv_index)
ks = [1.4, 1.5, 1.6]
vocab_sizes = [1500]
qts = ["Implicit semantic text retrieval and distributed implementation for rural medical care"
                ,"A systematic study of double auction mechanisms in cloud computing"
                ,"Memory Efficiency of Parallel Programs and Memory Bounded Speedup"
                ,"Computer-Aided Design of Machine Learning Algorithm: Training Fixed-Point Classifier for On-Chip Low-Power Implementation"
                ,"Web Service Clustering Using Relational Database Approach"
                ,"COMPSs-Mobile: Parallel Programming for Mobile-Cloud Computing"
                ,"On musical performances identification, entropy and string matching"
                ,"EDISON: A Web-Based HPC Simulation Execution Framework for Large-Scale Scientific Computing Software"
                ,"Learning to be efficient: algorithms for training low-latency, low-compute deep spiking neural networks"
                ,"The Internet of Things: Opportunities and Challenges for Distributed Data Analysis"
                ,"Derandomizing Local Distributed Algorithms under Bandwidth Restrictions"
                ,"Compact Implementations of LSH"
                ,"Fog and IoT: An Overview of Research Opportunities"
                ,"Virtualization layer security challenges and intrusion detection/prevention systems in cloud computing: a comprehensive review"
                ,"Image Segmentation via Probabilistic Graph Matching"
                ,"Empowering network operating systems with memory management techniques"
                ,"Recurrently Target-Attending Tracking"
                , "A mobile follow-me cloud content caching model"
                , "Adaptive Power Control for A-CDMA Systems Over K-Distributed Fading Channels"]
for k in ks:
    t = time.time()
    ri = inv_index.InvInd(bm25_k = k, top_words=1500)
    ri.preprocess_data("abstract")
    ri.build_vocab("abstract")
    ri.compute_IDF("abstract")
    ri.preprocess_data("title")
    ri.build_vocab("title")
    ri.compute_IDF("title")
    ri.inverted_index()
    t2 = time.time()
    print("Done with ii, starting testing")
    print(str((t2 - t)/60) + " minutes")
    for v in vocab_sizes:
        rs = recsys_ii.RecSys_II(bm25_k=k, top_words=v)
        rs.build_vocab("abstract")
        rs.build_vocab("title")
        reciprocal_rank = 0
        for qt in qts:
            ss = rs.similarity_ranking(qt, qt)
            top100_indices = np.argsort(ss)[-1000:][::-1]

            s0 = ss[top100_indices[0]]

            for i in range(1, 6):
                if ss[top100_indices[i]] == s0:
                    print("Scores are same for query: " + qt)
                    print("k is " + str(k) + " and vocab size is " + str(v))

            for rank, idx in enumerate(top100_indices, start=1):
                if rs.dataset["title"].iloc[idx] == qt:
                    reciprocal_rank += 1 / rank
                    break
        
        reciprocal_rank /= len(qts)
        with open("hp-k.txt", "a") as f:
            f.write("k: " + str(k) + "\n")
            f.write("vocab size: " + str(v) + "\n")
            f.write("MRR: " + str(reciprocal_rank) + "\n")
            f.write("-----------------------------------------------------------------------\n")
    t3 = time.time()
    print("Done with the iteration")
    print(str((t3 - t2)/60) + " minutes")
```
