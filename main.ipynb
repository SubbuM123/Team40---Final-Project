```python
# these are the given parameters that yielded the best results during testing
# bm25_k represents the k value in the Okapi-bm25 transformation for TFIDF
# top_words represents the vocab size for the model
# b represents the constant in document length normalization, that rewards and incentivizes shorted and longer documents
ri = inv_index.InvInd(bm25_k = 1, top_words=1500, b=0.75)

ri.preprocess_data("abstract")
ri.build_vocab("abstract")
ri.compute_IDF("abstract")
ri.preprocess_data("title")
ri.build_vocab("title")
ri.compute_IDF("title")
ri.inverted_index()
```

Write you query below, and run the cell below that to retrieve the top 5 most relevant documents.

```python
query = "Text retrieval models"
```

```python
rs = recsys_ii.RecSys_II(bm25_k=1, top_words=1500, b = 0.7)

rs.build_vocab("abstract")
rs.build_vocab("title")

ss = rs.similarity_ranking(query, query)
top5_indices = np.argsort(ss)[-5:][::-1]

print("Top 5 most relevant documents")
rank = 1
for idx in top5_indices:
    print(str(rank) + ". " + rs.dataset["title"].iloc[idx])
    rank += 1
```
